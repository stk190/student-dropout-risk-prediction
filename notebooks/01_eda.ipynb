```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("../data/data.csv", sep=";")

print(df.shape)

```

```python
df["Target"].value_counts()
```

```python
df["Target"].value_counts().plot(kind="bar")

plt.title("Target Distribution")
plt.xlabel("Target")
plt.ylabel("Number of Students")
plt.tight_layout()

plt.savefig("../figures/target_distribution.png")
plt.show()
```

```python
plt.hist(df["Age at enrollment"], bins=10)

plt.title("Age Distribution")
plt.xlabel("Age at Enrollment")
plt.ylabel("Number of Students")
plt.tight_layout()

plt.savefig("../figures/age_distribution.png")
plt.show()
```

```python
df.boxplot(column="Admission grade", by="Target")

plt.title("Admission Grade by Target")
plt.suptitle("")
plt.xlabel("Target")
plt.ylabel("Admission Grade")
plt.tight_layout()

plt.savefig("../figures/admission_grade_by_target.png")
plt.show()
```

```python
df.boxplot(column="Curricular units 1st sem (grade)", by="Target")

plt.title("First Semester Grade by Target")
plt.suptitle("")
plt.xlabel("Target")
plt.ylabel("First Semester Grade")
plt.tight_layout()

plt.savefig("../figures/first_semester_grade_by_target.png")
plt.show()
```

```python
df.boxplot(column="Curricular units 2nd sem (grade)", by="Target")

plt.title("Second Semester Grade by Target")
plt.suptitle("")
plt.xlabel("Target")
plt.ylabel("Second Semester Grade")
plt.tight_layout()

plt.savefig("../figures/second_semester_grade_by_target.png")
plt.show()
```

```python
df.boxplot(column="Curricular units 1st sem (approved)", by="Target")

plt.title("First Semester Approved Courses by Target")
plt.suptitle("")
plt.xlabel("Target")
plt.ylabel("Approved Courses")
plt.tight_layout()

plt.savefig("../figures/approved_courses_by_target.png")
plt.show()
```

```python
correlation = df.drop("Target", axis=1).corr()

plt.figure(figsize=(12, 10))
plt.imshow(correlation, aspect="auto")
plt.colorbar()

plt.title("Feature Correlation Matrix")
plt.tight_layout()

plt.savefig("../figures/correlation_matrix.png")
plt.show()
```
